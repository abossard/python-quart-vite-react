## 🎯 Wähl dein Modell

Über LiteLLM kannst du über 100 verschiedene Modelle ansprechen. Die verfügbaren Modelle werden automatisch aus deiner `.env`-Konfiguration erkannt.

```python
import ipywidgets as widgets
from IPython.display import display
from dspy_tasks.config import get_available_models, get_default_model, configure_dspy

# Available models (add/remove based on your setup)
AVAILABLE_MODELS = get_available_models()

model_dropdown = model_picker(AVAILABLE_MODELS, default=get_default_model())
display(model_dropdown)
```

## ✨ Deine erste Vorhersage

Eine **Signature** beschreibt was reingeht und was rauskommen soll — wie ein Funktionstyp. Kein Prompt nötig! Der passende Prompt wird automatisch generiert.

```python
from dspy_tasks.data import ClassifySentiment
from dspy_tasks.config import configure_dspy

# Configure DSPy with chosen model
configure_dspy(model=model_dropdown.value)

# This is ALL you need — no prompt engineering!
classifier = dspy.Predict(ClassifySentiment)

# Try it
result = classifier(review="This laptop is absolutely fantastic! Best purchase I've made all year.")
print(f"🎯 Sentiment: {result.sentiment}")
```

```python
review_input = widgets.Textarea(value="The battery life is terrible, but the screen quality is amazing.",
                                  description="Review:", layout=widgets.Layout(width="600px", height="80px"))
run_btn = widgets.Button(description="Classify!", button_style="primary", icon="play")
output = widgets.Output()

def on_click(b):
    with output:
        output.clear_output()
        result = classifier(review=review_input.value)
        display_score("Prediction", 1.0 if result.sentiment else 0.0)
        print(f"→ {result.sentiment}")

run_btn.on_click(on_click)
display(review_input, run_btn, output)
```
