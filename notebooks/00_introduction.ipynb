# 🚀 Der Paradigmenwechsel: Prompts als kompilierte Artefakte

> *"Das Ziel von DSPy ist es, Prompts wie kompilierte Binaries zu behandeln — du sagst WAS du willst, nicht WIE."*

Willkommen beim **DSPy Playground**! Diese Notebook-Serie zeigt dir einen fundamental anderen Ansatz, wie man mit LLMs arbeitet:

- **Kein Prompt Engineering mehr** — beschreibe *was* du willst, DSPy findet heraus *wie*
- **Messbare Qualität** — jede Aufgabe hat Metriken, jede Änderung wird evaluiert
- **Systematische Optimierung** — kompiliere deine Prompts wie Code, statt sie wie Gedichte zu basteln

Am Ende hast du 20 immer schwierigere Aufgaben durchgespielt, sie über verschiedene Modelle optimiert, und gesehen wie DSPy Prompt Engineering von Kunst in Ingenieurwesen verwandelt.

## 🎯 Wähl dein Modell

DSPy nutzt LiteLLM als Backend — damit kannst du über 100 verschiedene Modelle ansprechen. Die verfügbaren Modelle werden automatisch aus deiner `.env`-Konfiguration erkannt.

```python
import ipywidgets as widgets
from IPython.display import display
from dspy_tasks.config import get_available_models, get_default_model, configure_dspy

# Available models (add/remove based on your setup)
AVAILABLE_MODELS = get_available_models()

model_dropdown = model_picker(AVAILABLE_MODELS, default=get_default_model())
display(model_dropdown)
```

## ✨ Deine erste DSPy-Vorhersage

Eine **Signature** in DSPy ist wie ein Funktionstyp — sie beschreibt was reingeht und was rauskommen soll. Kein Prompt nötig! DSPy generiert den Prompt automatisch.

```python
from dspy_tasks.data import ClassifySentiment
from dspy_tasks.config import configure_dspy

# Configure DSPy with chosen model
configure_dspy(model=model_dropdown.value)

# This is ALL you need — no prompt engineering!
classifier = dspy.Predict(ClassifySentiment)

# Try it
result = classifier(review="This laptop is absolutely fantastic! Best purchase I've made all year.")
print(f"Sentiment: {result.sentiment}")
```

```python
review_input = widgets.Textarea(value="The battery life is terrible, but the screen quality is amazing.",
                                  description="Review:", layout=widgets.Layout(width="600px", height="80px"))
run_btn = widgets.Button(description="Classify!", button_style="primary", icon="play")
output = widgets.Output()

def on_click(b):
    with output:
        output.clear_output()
        result = classifier(review=review_input.value)
        display_score("Prediction", 1.0 if result.sentiment else 0.0)
        print(f"→ {result.sentiment}")

run_btn.on_click(on_click)
display(review_input, run_btn, output)
```

## 📋 Die 20 Aufgaben vor dir

Von einfacher Textklassifikation bis hin zu Agenten die Tools nutzen — jede Aufgabe hat eine klare Metrik und einen Lerneffekt.

```python
for tier in range(1, 5):
    tasks = [t for t in task_summary() if t["tier"] == tier]
    tier_names = {1: "Fundamentals", 2: "Reasoning", 3: "Composition", 4: "Agentic"}
    display_tier_header(tier, tier_names[tier], f"{len(tasks)} tasks")
    for t in tasks:
        print(f"  {'★' * tier}{'☆' * (4-tier)}  {t['name']} ({t['module_type']})")
        print(f"      {t['description']}")
        print()
```
